There are three loss-making sub-categories: Tables, Bookcases and Supplies. Tables has a profit margin of -8.56%.This means that, across the recorded transactions, the sub-category lost approximately 8.56 for every 100 in sales. It also has the largest total loss: 17,725.48.

There are five sub categories with highest margins: Labels, Paper, Envelopes, Copiers and Fasteners. Labels has the highest margin at 44.42%, but its total profit is only 5,546.25 because its sales volume is relatively low.

Business Question: Are Discounts Associated with Lower Profit?

```python
discount_analysis = df.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Discount=("Discount", "mean")
    )
```

```python
discount_analysis["Profit Margin %"] = (discount_analysis["Total_Profit"] / discount_analysis["Total_Sales"]) * 100
```

```python
discount_analysis = discount_analysis.sort_values(by="Average_Discount", ascending=False)
```

```python
print(discount_analysis.round(3))
```

Tables has the highest average discount from the three low-profitable sub-categories and the largest total loss. Bookcases also has a relatively high average discount and a negative profit margin. However, Supplies is among the low-profitable sub categoryits average discount is only 7.7%, yet it still generates an overall loss.

Binders has the highest average discount yet it generates a positive profit margin of 14.86%. Machines has an average discount of 30.6%, but its profit margin is only 1.79%. A high average discount does not automatically mean a product category will lose money. The outcome also depends on factors such as selling prices, costs, and the mix of products sold.

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=discount_analysis,
    x="Average_Discount",
    y="Profit Margin %",
    s=100
)

plt.title("Average Discount vs Profit Margin by Sub-Category")
plt.xlabel("Average Discount")
plt.ylabel("Profit Margin (%)")

plt.axhline(y=0, linestyle="--")
plt.tight_layout()
plt.show()
```

Business Question: Which customer segments generate the most sales and profit?

```python
segment_analysis = df.groupby("Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique"),
    Total_Customers=("Customer ID", "nunique")
)

segment_analysis["Profit Margin (%)"] = (
    segment_analysis["Total_Profit"] /
    segment_analysis["Total_Sales"]
) * 100

print(
    segment_analysis.sort_values(
        by="Total_Sales",
        ascending=False
    ).round(2)
)
```

Consumer is the largest segment. Consumer customers generate the most total sales and profit, but Home Office customers have the highest profit margin. The company should understand what drives Consumer sales and investigate whether successful products or sales strategies could be used to support other segments.

```python
print("First Order Date:", df["Order Date"].min())
print("Last Order Date:", df["Order Date"].max())
```

```python
df["Year"] = df["Order Date"].dt.year

yearly_analysis = df.groupby("Year").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
)

yearly_analysis["Profit Margin (%)"] = (
    yearly_analysis["Total_Profit"] /
    yearly_analysis["Total_Sales"]
) * 100

print(yearly_analysis.round(2))
```

Sales generally increased over the four-year period after a small decline in 2015. The highest annual sales occurred in 2017. The profit margin increased from 10.24% in 2014 to 13.10% in 2015, reaching its highest point of 13.43% in 2016. It then decreased slightly to 12.74% in 2017. Although 2017 had the highest total sales and profit, 2016 had the highest profit margin. This means the business retained a slightly larger proportion of its sales as profit in 2016.

The business processed considerably more orders in 2017 than in 2014, indicating growth in order volume. However, order counts alone cannot establish whether customer satisfaction or customer retention improved.

```python
yearly_analysis["Total_Sales"].plot(
    kind="line",
    marker="o",
    figsize=(9, 5),
    title="Yearly Sales Trend (2014–2017)"
)

plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.xticks(yearly_analysis.index)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
yearly_analysis["Total_Profit"].plot(
    kind="bar",
    figsize=(9, 5),
    title="Yearly Profit (2014–2017)"
)

plt.xlabel("Year")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
yearly_analysis["Profit Margin (%)"].plot(
    kind="line",
    marker="o",
    figsize=(9, 5),
    title="Yearly Profit Margin (2014–2017)"
)

plt.xlabel("Year")
plt.ylabel("Profit Margin (%)")
plt.xticks(yearly_analysis.index)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
yearly_analysis["Sales Growth (%)"] = (
    yearly_analysis["Total_Sales"].pct_change() * 100
)

yearly_analysis["Profit Growth (%)"] = (
    yearly_analysis["Total_Profit"].pct_change() * 100
)

print(yearly_analysis.round(2))
```

Analysing Monthly Sales Trends

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month_Name"] = df["Order Date"].dt.month_name()

print(df[["Order Date", "Year", "Month", "Month_Name"]].head())
```

```python
monthly_sales = df.groupby("Month").agg(
    Total_Sales=("Sales", "sum"),
    Total_Orders=("Order ID", "nunique"),
    Total_Profit=("Profit", "sum")
)

monthly_sales["Month_Name"] = monthly_sales.index.map(
    lambda month: pd.Timestamp(month=month, day=1, year=2000).month_name()
)

print(monthly_sales.round(2))
```

November is the strongest month for sales. December generates the highest total profit. February is the weakest month overall.

```python
monthly_sales.plot(
    y="Total_Sales",
    kind="bar",
    figsize=(12, 6),
    legend=False,
    title="Total Sales by Month (2014–2017)"
)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(
    ticks=range(12),
    labels=monthly_sales["Month_Name"],
    rotation=45
)
plt.tight_layout()
plt.show()
```

```python
yearly_monthly_sales = df.pivot_table(
    index="Month",
    columns="Year",
    values="Sales",
    aggfunc="sum"
)

yearly_monthly_sales.index = yearly_monthly_sales.index.map(
    lambda month: pd.Timestamp(
        year=2000, month=month, day=1
    ).month_name()
)

print(yearly_monthly_sales.round(2))
```

```python
yearly_monthly_sales.plot(
    kind="line",
    marker="o",
    figsize=(12, 6),
    title="Monthly Sales Trends by Year"
)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(12), yearly_monthly_sales.index, rotation=45)
plt.legend(title="Year")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

September, November, and December are promising months to investigate for recurring high sales. February appears to be a consistently weaker month in the combined totals, but we should verify its year-by-year performance before drawing a firm conclusion.

```python
average_monthly_sales = yearly_monthly_sales.mean(axis=1)

average_monthly_sales = average_monthly_sales.sort_values(
    ascending=False
)

print(average_monthly_sales.round(2))
```

```python
average_monthly_sales.plot(
    kind="bar",
    figsize=(12, 6),
    title="Average Monthly Sales (2014–2017)"
)

plt.xlabel("Month")
plt.ylabel("Average Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_growth = yearly_monthly_sales.pct_change() * 100

print(monthly_growth.round(2))
```

March is a strong sales month relative to February, but the percentage increase should be interpreted alongside the actual sales values. September appears to be a recurring strong month in this dataset. ShopSmart could use this finding to inform inventory and promotional planning, subject to further validation.Sales increased from October to November in all four years.This supports your earlier finding that November has the highest average monthly sales. October appears to be a recurring slowdown after September's relatively strong sales. ShopSmart could investigate product demand and promotional activity during this period. December sales declined in 2014, 2015, and 2017, but increased in 2016. Although December had the second-highest average monthly sales, it did not consistently outperform November.

```python
year_over_year_monthly_growth = (
    yearly_monthly_sales.pct_change(axis="columns") * 100
)

print(year_over_year_monthly_growth.round(2))
```

The business experienced a strong January in 2017. August sales were particularly strong in 2017, although the year-to-year variation suggests that August performance was not consistent across the entire period. November sales were relatively stable between 2014 and 2016, then increased by 49.16% in 2017. This reinforces the earlier finding that November is the strongest month by average sales. March sales decreased by 30.46% in 2015, then increased by 33.54% in 2016 and 13.84% in 2017. March sales recovered after the decline in 2015 and continued growing in the following two years. December sales increased in 2015 and 2016, but declined by 13.58% in 2017. A month can have a high average sales value and still experience a decline in a particular year.

```python
plt.figure(figsize=(12, 7))

sns.heatmap(
    yearly_monthly_sales,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu"
)

plt.title("Monthly Sales by Year (2014–2017)")
plt.xlabel("Year")
plt.ylabel("Month")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 7))

sns.heatmap(
    year_over_year_monthly_growth,
    annot=True,
    fmt=".1f",
    cmap="RdYlGn",
    center=0
)

plt.title("Year-over-Year Monthly Sales Growth (%)")
plt.xlabel("Year")
plt.ylabel("Month")
plt.tight_layout()
plt.show()
```
